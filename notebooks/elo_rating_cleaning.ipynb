```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/elo_setup_new.csv")
```

```python
df.info() 
```

```python
df.head(5)
```

```python
df = df.drop(columns = ["host", "team_a_rank_change", "team_b_rank_change", "points"])
```

```python
df.isnull().sum() 
```

```python
df[df.isna().any(axis=1)]
```

```python
cols = [
    "date",
    "team_a",
    "team_b",
    "team_a_rank",
    "team_b_rank"
]

null_rows = df[cols].isnull().any(axis=1)

print(null_rows.sum())
display(df.loc[null_rows].head(20) )
```

```python
df = df.dropna(subset = ["team_a", "team_a_rank", "team_b_rank", "team_b", "date"])
df.isnull().sum()
```

```python
df["date"] = pd.to_datetime(df["date"])

categorical = [
    "tournament",
    "team_a",
    "team_b"
]

for col in categorical:
    df[col] = df[col].astype("category")
df.info()
```
