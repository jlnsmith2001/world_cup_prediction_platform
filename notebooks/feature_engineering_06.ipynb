```python
import pandas as pd
import numpy as np 
```

```python
matches = pd.read_csv("../data/processed/matches_elo_merged_processed.csv")
```

```python
matches.info() 
```

```python
matches = matches.drop(columns = "Unnamed: 0")
```

```python
matches["elo_difference"] = (matches["home_elo"] -  matches["away_elo"])

matches["rank_difference"] = (matches["away_ranking"] - matches["home_ranking"])

matches["home_win"] = np.where(matches["home_score"] > matches["away_score"], 1, 0)
matches["home_loss"] = np.where(matches["home_score"] < matches["away_score"], 1, 0)
matches["home_draw"] = np.where(matches["home_score"] == matches["away_score"], 1, 0)

matches["away_win"] = np.where(matches["away_score"] > matches["home_score"], 1, 0)
matches["away_loss"] = np.where(matches["away_score"] < matches["home_score"], 1, 0)
matches["away_draw"] = np.where(matches["home_score"] == matches["away_score"], 1, 0)
```

```python
matches.tail(10) 
```

```python
matches[[
    "home_team", "home_score",
    "away_team", "away_score", 
    "home_win", "home_loss",
    "home_draw", "away_win",
    "away_loss", "away_draw"
]].tail(15) 
```

```python
matches_helper_home = matches.copy()
matches_helper_away = matches.copy() 

matches_helper_home = matches_helper_home.drop(columns=[
    "city", "country", "neutral", "tournament_code", 
    "tournament", "home_elo", "away_elo", "home_ranking",
    "away_ranking", "away_win", "away_loss", "away_draw",
    "elo_difference", "rank_difference"])

matches_helper_away = matches_helper_away.drop(columns=[
    "city", "country", "neutral", "tournament_code", 
    "tournament", "home_elo", "away_elo", "home_ranking",
    "away_ranking", "home_win", "home_loss", "home_draw",
    "elo_difference", "rank_difference"])


matches_helper_home = matches_helper_home.rename(columns={"home_team": "team", "away_team": "opponent", "home_score": "goals_for", 
                                                          "away_score": "goals_against", "home_win": "win", 
                                                          "home_draw": "draw", "home_loss": "loss"})


matches_helper_away = matches_helper_away.rename(columns={"away_team": "team", "home_team": "opponent", "away_score": "goals_for", 
                                                          "home_score": "goals_against", "away_win": "win", 
                                                          "away_draw": "draw", "away_loss": "loss"})


matches_helper_home.tail(5) 
```

```python
matches_helper_away.tail(5) 
```

```python
print(matches_helper_away.shape)
print(matches_helper_home.shape) 
```

```python
matches_helper_combined = pd.concat([matches_helper_home, matches_helper_away], ignore_index  = True)

matches_helper_combined = matches_helper_combined.sort_values(by=["team", "date"])

matches_helper_combined.tail(20) 
```

```python
matches_helper_combined.head(5)
```

```python
matches_helper_combined["prev_win"] = matches_helper_combined.groupby(["team"])["win"].shift(1)
matches_helper_combined["last_5_win_rate"] = matches_helper_combined.groupby(["team"])["prev_win"].rolling(window=5).mean().reset_index(level=0, drop = True)


matches_helper_combined["prev_loss"] = matches_helper_combined.groupby(["team"])["loss"].shift(1)
matches_helper_combined["last_5_loss_rate"] = matches_helper_combined.groupby(["team"])["prev_loss"].rolling(window=5).mean().reset_index(level=0, drop = True)

matches_helper_combined["prev_goals_for"] = matches_helper_combined.groupby(["team"])["goals_for"].shift(1)
matches_helper_combined["last_5_scoring_rate"] = matches_helper_combined.groupby(["team"])["prev_goals_for"].rolling(window=5).mean().reset_index(level=0, drop = True)


matches_helper_combined["prev_goals_against"] = matches_helper_combined.groupby(["team"])["goals_against"].shift(1)
matches_helper_combined["last_5_conceding_rate"] = matches_helper_combined.groupby(["team"])["prev_goals_against"].rolling(window=5).mean().reset_index(level=0, drop = True)

```

```python
matches_helper_combined.loc[matches_helper_combined["team"] == "France", ["date", "team", "opponent", "win", "last_5_win_rate", "last_5_loss_rate", "last_5_scoring_rate", "last_5_conceding_rate"]].tail(20) 
```
